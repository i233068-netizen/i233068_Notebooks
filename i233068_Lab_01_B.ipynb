```python
num_list = [2, 5, 7, 8, 10, 3]

even_squares = []
odd_cubes = []

for n in num_list:
    if n % 2 == 0:
        even_squares.append(n**2)
    else:
        odd_cubes.append(n**3)

merged_list = sorted(even_squares + odd_cubes)

print("Even Squares:", even_squares)
print("Odd Cubes:", odd_cubes)
print("Merged Sorted:", merged_list)
```

```python
numbers = [10, 50, 100, 80, 30]

abs_difference = lambda a,b: abs(a-b)

min_value = min(numbers)
max_value = max(numbers)

normalize = lambda x: (x-min_value)/(max_value-min_value)

normalized = list(map(normalize, numbers))
filtered = list(filter(lambda x: x>0.5, normalized))

print("Normalized:", normalized)
print("Filtered (>0.5):", filtered)
print("Absolute Difference:", abs_difference(10,50))
```

```python
def flatten(list):
    result_array=[]
    for i in list:
        if type(i)==list:
            result_array.extend(flatten(i))
        else:
            result_array.append(i)
    return result_array

house_extensions = ['bed',112.0,'hall',113.5,'bath',189.5,'kit',789.0,'liv',230.0,
         ['pool',112.7,117.9],['grg',112.0]]

print(flatten(house_extensions))
```

```python
Fav_food_items = [
 ['pizza','burger','hotdogs'],
 ['pasta','hotdogs'],
 ['pizza'],
 ['burger','hotdogs'],
 ['rice','pasta'],
 ['pasta']
]

def is_subset(a,b):
    for item in a:
        if item not in b:
            return False
    return True

result_array=[]

for i in range(len(Fav_food_items)):
    subset=False
    for j in range(len(Fav_food_items)):
        if i!=j and is_subset(Fav_food_items[i],Fav_food_items[j]):
            subset=True
    if not subset:
        result_array.append(i)

print(result_array)
```

```python
resource_usage = [
{'cpu':45,'memory':70},
{'cpu':85,'memory':90},
{'cpu':60,'memory':40},
{'cpu':95,'memory':88},
{'cpu':30,'memory':20}
]

valid=[u for u in resource_usage if u['cpu']>80 and u['memory']>75]

cpu_values=[u['cpu'] for u in resource_usage]
memory_values=[u['memory'] for u in resource_usage]

normalized=lambda x,minv,maxv:(x-minv)/(maxv-minv)

scores_array=[]
for u in resource_usage:
    c=normalized(u['cpu'],min(cpu_values),max(cpu_values))
    m=normalized(u['memory'],min(memory_values),max(memory_values))
    scores_array.append(c+m)

rank=np.argsort(scores_array)[::-1]

print("Valid:",valid)
print("Ranking:",rank)
```

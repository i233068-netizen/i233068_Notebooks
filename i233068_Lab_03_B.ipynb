```python
class SmartVacuumAgent:
    def __init__(self):
        self.internal_state = {"RoomA": None, "RoomB": None}

    def decide_action(self, perception):
        position, condition = perception
        self.internal_state[position] = condition

        if condition == "Dirty":
            return "CLEAN"
        elif position == "RoomA":
            return "MOVE_RIGHT"
        else:
            return "MOVE_LEFT"


world = {"RoomA": "Dirty", "RoomB": "Clean"}
vacuum_agent = SmartVacuumAgent()

for step in range(4):
    current_room = "RoomA" if step % 2 == 0 else "RoomB"
    sensed_input = (current_room, world[current_room])
    chosen_action = vacuum_agent.decide_action(sensed_input)

    if chosen_action == "CLEAN":
        world[current_room] = "Clean"

    print(f"Time Step {step + 1}")
    print("Perception:", sensed_input)
    print("Action Taken:", chosen_action)
    print("Agent Memory:", vacuum_agent.internal_state)
    print()
```

```python
routes = {
    "Route A": {"distance": 10, "time": 15, "risk": "Low"},
    "Route B": {"distance": 8, "time": 20, "risk": "Medium"},
    "Route C": {"distance": 12, "time": 10, "risk": "High"}
}

risk_mapping = {
    "Low": 1,
    "Medium": 2,
    "High": 3
}

# Step 3: Define utility function
# Higher utility = better route
def utility_function(distance, time, risk):
    return (1 / distance) + (1 / time) + (1 / risk)

# Step 4: Compute utilities
utility_scores = {}

for route, data in routes.items():
    d = data["distance"]
    t = data["time"]
    r = risk_mapping[data["risk"]]

    utility_scores[route] = utility_function(d, t, r)

# Step 5: Select best route
best_route = max(utility_scores, key=utility_scores.get)

# Step 6: Display results
print("Utility Scores:")
for route, score in utility_scores.items():
    print(f"{route}: {score:.4f}")

print("\nSelected Route:")
print(best_route)
```
